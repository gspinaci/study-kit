## Setup: synthetic data in matrix form (§1.1 · ch01-o1)

Rows of `X` are samples, columns are features; a column of ones gives the intercept.

```python
import numpy as np

rng = np.random.default_rng(seed=0)
n, d = 50, 3
features = rng.normal(size=(n, d))
X = np.column_stack([np.ones(n), features])
w_true = np.array([1.0, 2.0, -1.0, 0.5])
y = X @ w_true + rng.normal(scale=0.3, size=n)
X.shape, y.shape
```

## From scratch: the normal equations (§1.3 · ch01-o2)

Solve \(X^\top X w = X^\top y\) with a linear solver (no explicit inverse).

```python
w_normal = np.linalg.solve(X.T @ X, X.T @ y)
residual = y - X @ w_normal
w_normal
```

## Geometry: the residual is orthogonal to the columns (§1.4 · ch01-o3)

```python
print('X^T r =', X.T @ residual)
assert np.allclose(X.T @ residual, 0, atol=1e-9)
```

## Library comparison: `lstsq` (§1.5 · ch01-o4)

`lstsq` works on an orthogonal factorization of X instead of forming \(X^\top X\).

```python
w_lstsq, *_ = np.linalg.lstsq(X, y, rcond=None)
print(np.abs(w_lstsq - w_normal).max())
assert np.allclose(w_lstsq, w_normal)
```

### Ill-conditioning (§1.5 · ch01-o4)

Make two columns nearly identical and compare condition numbers.

```python
X_bad = X.copy()
X_bad[:, 2] = X_bad[:, 1] + 1e-6 * rng.normal(size=n)
print(f'cond(X)     = {np.linalg.cond(X_bad):.2e}')
print(f'cond(X^T X) = {np.linalg.cond(X_bad.T @ X_bad):.2e}  # about the square')
```
